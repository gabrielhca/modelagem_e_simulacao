## Monte Carlo - dado $d_{20}$

No RPG Dungeons & Dragons, para superar um desafio o jogador deve lançar um dado de 20 faces ($d_{20}$) e comparar o resultado a um
número alvo. Em certas situações, o jogador ganha um benefício chamado vantagem, e pode lançar dois $d_{20}$, escolhendo em seguida o
maior valor. Qual o valor esperado de um lançamento sem vantagem e com vantagem?

#### Valores para 1 dado

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

face = rng.integers(1, 21, size=10_000_000)
#print(face)

# histograma com distribuição dos valores
plt.hist(face, bins=20)
plt.show()
```

```python
print('Valor esperado: ', np.average(face))
```

#### Valores para 2 dados

```python
dado1 = rng.integers(1, 21, size=10_000_000)
dado2 = rng.integers(1, 21, size=10_000_000)
maiorvalor = np.maximum(dado1, dado2)

# histograma com distribuição dos valores
plt.hist(maiorvalor, bins=20)
plt.show()
```

```python
print('Valor esperado: ', np.average(maiorvalor))
```
