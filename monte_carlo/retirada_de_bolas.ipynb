## Monte Carlo - retirada de bolas brancas e pretas

Suponha que temos uma caixa com 5 bolas pretas e 7 bolas brancas. Retiramos duas sem reposição. Qual é a distribuição de probabilidade da variável aleatória X = número de bolas pretas retiradas? E qual é o número esperado de bolas pretas retiradas?

```python
import numpy as np
import matplotlib.pyplot as plt

# gerador de numeros aleatórios
rng = np.random.default_rng()

nsim = 1_000_000
res = np.zeros(nsim) # array com zeros

for i in range(nsim):
  npretas = 5
  nbrancas = 7

  for _ in range(2): # 2 sorteios
    if rng.integers(1, npretas + nbrancas + 1) <= npretas:
      res[i] += 1
      npretas -= 1
    else:
      nbrancas -= 1

# histograma com distribuição dos valores
plt.hist(res, bins=3, density=True)
plt.show()
```
