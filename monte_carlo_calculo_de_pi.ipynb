## Monte Carlo - Cálculo de $\pi$

```python
import numpy as np
rng = np.random.default_rng()

N = 10_000_000
n = 0

x = rng.uniform(size=N)
y = rng.uniform(size=N)

for i in range(N):
    if x[i]**2 + y[i]**2 < 1:
        n+=1

pi = 4*n/N
pi
```

segunda versão: gráfico

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

N = 10000
n = 0

x = rng.uniform(size=N)
y = rng.uniform(size=N)
pi = []

for i in range(N):
    if x[i]**2 + y[i]**2 < 1:
        n+=1
    pi.append(4*n/(i+1)) # parcial
plt.plot(range(i+1),pi)
plt.axhline(y=np.pi, color='r', linestyle='-')
plt.xlabel('Número de pontos')
plt.ylabel('Aproximação')
plt.show()
```
